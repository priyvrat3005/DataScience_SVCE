## Navie Bayes Algorithm 
Naive Bayes is a supervised Machine Learning Algorithm basically used for classifications like spam detection ,disease prediction sentiment analysis etc.

#### Types of Naive Bayes Classification Algos :
   ##### 1. Gaussaian Naive Bayes - when we want to predict some based of on the continuous data like disease prediction salary blood pressure (these data are conitinuos)
   ##### 2. MultiNomial Naive Bayes - when the features represents discrete counts or frequencies (lets suppose in a
   ###### mail how many times "free" or "win" words appear )
   ##### 3. Bernoulli Naive Bayes 

### Algo            |||                 Feature              |||                         Use Case 
### Gaussian        |||         continuous /numerical        |||       Iris flower type pred,predicting the house price 
### MultiNomial     |||         Discrete count/ Frequencies   |||    Product categorization / Text analysis / Spam detection
### Bernoulli                 Binary/Boolean Flags          word exist or not success /failure prediction

##### Task : We are going to predict whether a mail is Spam or Ham

```python
from sklearn.feature_extraction.text import  CountVectorizer 
from sklearn.naive_bayes import MultinomialNB  
```

```python
## This is our training data for the model but in text form 
emails=["Get an free lottery cash prize now",
        "win millions in online lottery",
        "Hey let's meet tomorrow for lunch",
        "please review the project report attached"]  # This is our feature 

labels=[1,1,0,0] # response  Y_train
```

##### In the below cell im gouing to convert the textual data in to numeric values using 
##### using the CountVectorizer algo

```python
vectorizer=CountVectorizer()
X_train=vectorizer.fit_transform(emails)

model=MultinomialNB()  # created the model of NaiveBayes 
model.fit(X_train,labels)
```

```python
new_mail=["Hey you gotta lottery of 500000"] # X_test
new_mail_ham=["I have shared the meeting link"]
X_test=vectorizer.transform(new_mail_ham)

pred=model.predict(X_test)

if pred[0]==1:
    print("The mail is spam")
else:
    print("The mail is Ham")
```
