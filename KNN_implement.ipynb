```python
import math 
```

```python
### Here we have defined the function to calculate the distance between two points 
def distance(pointa,pointb):
    sumsrqd=0
    for i in range(len(pointa)) :
        diff=pointa[i]-pointb[i]
        sumsrqd=sumsrqd+(diff**2)
    return math.sqrt(sumsrqd)


```

```python

# train_data=[[2,10],[5,20],[18,7]]
#     #        0       1      2
# labels=["cat","cat","Lion"]
#         #  0     1     2 
# test_point=[9,7]

def kneighbors(train_data,labels,test_point,k):
    distances=[]
    for index,point in enumerate(train_data):
        dist=distance(test_point,point)
        distances.append((dist,index,point))

    neighbors=[] # this variable will store dist point and labels 
    for i in range(k):
        dist,indx,point=distances[i]
        lbl=labels[indx]
        neighbors.append((point,lbl,dist))
    return neighbors

# kneighbors(train_data,labels,test_point,3) ## herw we have called the function 
```

```python
# d=[([2, 10], 'cat', 7.615773105863909),
#  ([5, 20], 'cat', 13.601470508735444),
#  ([18, 7], 'Lion', 9.0)]

def majority(nbrs):
    labl_count={}

    for _,lbl,_ in nbrs:
        if lbl not in labl_count:
            labl_count[lbl]=0
        labl_count[lbl]+=1
    
    max_labl=None
    max_count=-1

    for label ,count in labl_count.items():
        if count>max_count:
            max_count=count
            max_labl=label
    return max_labl ,max_count


# majority(d)
```

```python
def KNpredict(train_data,labels,new_point,k):
    neighbors=kneighbors(train_data,labels,new_point,k)
    result=majority(neighbors)
    return result , neighbors
```
